## 1. Data Loading
We use a synthetic dataset designed for demonstration, containing English and Roman Urdu reviews.

```python
df = load_sample_dataset()
print(f"Loaded {len(df)} reviews.")
df.head()
```

## 2. Text Preprocessing
Cleaning URLs, special characters, and normalizing Roman Urdu.

```python
sample = "Muje yeh product bht pasand aaya! 10/10 😍 check it out: https://xyz.com"
print(f"Original:  {sample}")
print(f"Processed: {preprocess_pipeline(sample)}")
```

## 3. Sentiment Analysis (ML Model)
Training a Multinomial Naive Bayes model on the reviews.

```python
sent_model = MLSentimentModel()
sent_model.train(df)

test_review = "The quality is absolutely amazing, bohot acha product hai."
prediction = sent_model.predict(test_review)
print(f"Review: {test_review}\nSentiment: {prediction}")
```

## 4. Intent Classification
Identifying if the customer wants a refund, has a delivery issue, or just a query.

```python
intent_classifier = IntentClassifier()
intent_classifier.train(df)

refund_review = "Mujhe wapis paise chahiye, item broke during shipping."
print(f"Review: {refund_review}\nIntent: {intent_classifier.predict(refund_review)}")
```

## 5. Topic Modeling (Unsupervised)
Discovering themes in the reviews using Non-Negative Matrix Factorization (NMF).

```python
tm = TopicModeler(n_topics=3)
tm.fit(df['review'].tolist())

print("Discovered Topics:")
for topic, words in tm.get_topics(5).items():
    print(f"{topic}: {', '.join(words)}")
```
